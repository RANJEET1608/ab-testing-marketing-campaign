```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Before campaign
before = np.random.normal(loc=50, scale=10, size=1000)

# After campaign (Slightly better)
after = np.random.normal(loc=52, scale=10, size= 1000)

df = pd.DataFrame({
    "group": ["Before"]*1000 + ["After"]*1000,
    "engagement": np.concatenate([before, after])
})

df.groupby("group")["engagement"].mean()
```

```python
from scipy.stats import ttest_ind

stats, p_values = ttest_ind(after, before)

if p_values < 0.05:
    print("The new marketing campaign significantly improved user engagement. The result is statistically significant, so the campaign can be implemented.")
else:
    print("The new marketing campaign did not show a statistically significant improvement in engagement.")
```
